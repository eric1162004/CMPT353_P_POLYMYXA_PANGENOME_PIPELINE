```python
from sklearn.cluster import AgglomerativeClustering
from src.pca_analysis import run_pca_analysis
from IPython.display import Image, display
import os

print("RUNNING AGGLOMERATIVE HIERARCHICAL CLUSTERING (4D):")

# 1. Fit Hierarchical Clustering on your 4D latent features
# 'average' linkage mimics the UPGMA trees standard in genomic analyses
hca = AgglomerativeClustering(n_clusters=4, linkage="complete")
hca_labels = hca.fit_predict(latent_features.values)

# 2. Create a clean DataFrame to match the contract your pipeline expects
import pandas as pd
df_hca_results = pd.DataFrame(index=latent_features.index)
df_hca_results["Cluster"] = hca_labels

print("\nHIERARCHICAL CLUSTER ASSIGNMENTS:")
display(df_hca_results.head())

# Print the size of each cluster to see if it isolated the small groups
print("\nCLUSTER SIZES DETECTED:")
print(df_hca_results["Cluster"].value_counts().to_string())

print("\nPROJECTING 4D Space VIA DYNAMIC PCA MODULE:")
hca_pca_plot_path = "plots/autoencoder_4d_projected_hca.png"
os.makedirs(os.path.dirname(hca_pca_plot_path), exist_ok=True)

# 3. Route your existing 4D CSV into the PCA tool along with the new HCA labels
run_pca_analysis(
    matrix_path=nn_4d_csv_path,
    plot_path=hca_pca_plot_path,
    cluster_labels=df_hca_results["Cluster"],
)

if os.path.exists(hca_pca_plot_path):
    print("\n2D PCA PROJECTION OF HIERARCHICAL CLUSTERING:")
    display(Image(filename=hca_pca_plot_path))
```
